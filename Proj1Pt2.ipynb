```python
from pathlib import Path

project = Path("Project1Part2")
project.mkdir(exist_ok=True, parents=True)

print("Project location", project.resolve())
```

```python
app_code= """
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():
    return '''
    <h1>Local Colab Container Demo</h1>
    <p>This application is running inside a Docker container.</p>
    '''

@app.route("/health")
def health():
    return {
        "status": "healthy",
        "containerized": True
    }

if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5000)
"""

# Assuming 'project' is defined in a previous cell as Path("Project1Part2")
from pathlib import Path
project = Path("Project1Part2")

with open(project / "app.py", "w") as f:
    f.write(app_code)
```

```python
requirements = """
Flask ==3.1.2
"""

with open("Project1Part2/requirements.txt", "w") as f:
    f.write(requirements.strip())
```

```python
dockerfile = """
FROM python:3.12-slim

WORKDIR /app

COPY requirements.txt .

RUN pip install --no-cache-dir -r requirements.txt

COPY app.py .

EXPOSE 5000

CMD ["python", "app.py"]
"""

with open("Project1Part2/Dockerfile", "w") as f:
    f.write(dockerfile.strip())
```
